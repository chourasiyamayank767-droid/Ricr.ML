```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from statsmodels.stats.outliers_influence import (
    variance_inflation_factor
)
```

```python
data=({
    "Area_sqft": [1000, 1500, 2000, 2500, 3000, 3500],
    "Area_sqm": [92.9, 139.4, 185.8, 232.3, 278.7, 325.2],
    "Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [15, 10, 8, 5, 3, 2]
})

df=pd.DataFrame(data)
```

```python
# Step 3: Calculate correlation matrix
correlation_matrix = df.corr()
```

```python
df
```

```python
# Step 4: Calculate VIF
x=df.copy()
```

```python
vif_data=pd.DataFrame()
vif_data['features']=x.columns
vif_data['VIF']=[
    variance_inflation_factor(x.values,i)for i in range(x.shape[1])
]
```

```python
# step 5 display vif
print('\n VIF result:')   
vif_data.round(2)
```

```python
data=({
    "Area_sqft": [1000, 1500, 2000, 2500, 3000, 3500],
    'floor': [2,3,2,4,3,2],
    "Bedrooms": [2, 3, 3, 4, 4, 5],
    "Age": [15, 10, 8, 5, 3, 2],
    "Price": [30, 36, 45, 55, 65, 32]
})
```

```python
import statsmodels.api as sm
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]

x=sm.add_constant(x)
```

```python
model=sm.OLS(y,x).fit()
print(model.summary())
```
